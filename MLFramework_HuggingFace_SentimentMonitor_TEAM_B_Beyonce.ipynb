## Problem Statement 1: Digital Wallet Feedback Analyzer
A digital wallet company receives large amounts of user feedback after transactions. The company wants to automatically identify whether users are satisfied or dissatisfied with the wallet experience and flag negative feedback for investigation.

```python
wallet_feedback = [
    "The money transfer was instant and the interface is very easy to use.",
    "My wallet balance did not update even after the payment was completed.",
    "Adding money through UPI was extremely convenient.",
    "The app froze while I was trying to make an urgent payment.",
    "I love how quickly I can scan and pay using this wallet.",
]

results = sentiment(wallet_feedback)

for text, r in zip(wallet_feedback, results):
    action = "INVESTIGATE" if r["label"] == "NEGATIVE" else "NORMAL"
    print(f"{action:12} | {r['label']:8} {r['score']:.2f} | {text}")
```

## Problem Statement 2: Stock Trading App Sentiment Monitor
A stock trading platform wants to understand investor reactions to its services. Reviews mentioning slow order execution, confusing charts, or successful trades should be classified automatically so that the company can identify areas requiring improvement.

```python
trading_reviews = [
    "My buy order was executed instantly and the portfolio updated immediately.",
    "The app crashed just when I was trying to sell my shares.",
    "The new profit and loss dashboard is very helpful.",
    "Charts take forever to load during market hours.",
    "The trading interface is simple and easy to understand.",
]

results = sentiment(trading_reviews)

for text, r in zip(trading_reviews, results):
    print(f"{r['label']:8} {r['score']:.2f} | {text}")

negative_count = sum(r["label"] == "NEGATIVE" for r in results)
print(f"\nNegative reviews: {negative_count}/{len(results)}")
```

## Problem Statement 3: Insurance Claim Experience Analyzer
An insurance company wants to analyze customer comments about the claim process. Positive feedback can indicate efficient service, while negative feedback can reveal delays, communication problems, or dissatisfaction with claim settlements.

```python
claim_feedback = [
    "My health insurance claim was processed quickly and without any problems.",
    "I have been waiting for weeks with no update about my claim.",
    "The customer service executive explained the process very clearly.",
    "My claim was rejected and nobody properly explained why.",
    "The reimbursement reached my account earlier than expected.",
]

results = sentiment(claim_feedback)

for text, r in zip(claim_feedback, results):
    if r["label"] == "NEGATIVE" and r["score"] >= 0.85:
        action = "PRIORITY REVIEW"
    else:
        action = "REGULAR"

    print(f"{action:15} | {r['label']:8} {r['score']:.2f} | {text}")
```

## Problem Statement 6 : Financial Customer Email Prioritizer

A financial services company receives hundreds of customer emails every day. Instead of manually reading every email, the company wants to use sentiment analysis to identify highly negative messages that may require faster attention from the support team.

```python


emails = [
    "Thank you for resolving my account issue so quickly.",
    "I have contacted support three times and nobody has solved my problem.",
    "I am happy with the service and appreciate your quick response.",
    "My account has been charged incorrectly and I want this fixed immediately.",
    "The representative was helpful and explained everything clearly.",
    "This is extremely frustrating and I still have no solution.",
]

results = sentiment(emails)

for i, (text, r) in enumerate(zip(emails, results), start=1):

    if r["label"] == "NEGATIVE" and r["score"] >= 0.85:
        priority = "URGENT"
    elif r["label"] == "NEGATIVE":
        priority = "REVIEW"
    else:
        priority = "NORMAL"

    print(
        f"Email {i} | {priority:7} | "
        f"{r['label']:8} {r['score']:.2f} | {text}"
    )
```
