```python


from langchain_text_splitters import RecursiveCharacterTextSplitter



loan_rules = """
Rule 1: Applicants must be at least 21 years old to apply for a personal loan.

Rule 2: The maximum age at loan maturity should generally not exceed 60 years for salaried applicants.

Rule 3: The applicant should have a stable source of income to demonstrate repayment capacity.

Rule 4: Salaried applicants should have a minimum monthly income of Rs 25,000.

Rule 5: Self-employed applicants should have a stable business income for at least 2 years.

Rule 6: Applicants with a credit score of 750 or above are considered low-risk applicants.

Rule 7: Applicants with a credit score between 650 and 749 may be considered after additional risk assessment.

Rule 8: Applicants with a credit score below 650 require enhanced review and may be rejected depending on other factors.

Rule 9: The applicant's total existing monthly loan obligations should generally not exceed 40% of monthly income.

Rule 10: Applicants with a debt-to-income ratio above 50% should normally not be approved for a standard personal loan.

Rule 11: Applicants with recent loan defaults require additional verification before approval.

Rule 12: Applicants with an active loan settlement or write-off record should undergo enhanced credit assessment.

Rule 13: Employment stability should be considered while evaluating salaried applicants.

Rule 14: Applicants who have been employed for less than 6 months may require additional income or employment verification.

Rule 15: Bank statements may be reviewed to verify regular income and identify unusual financial activity.

Rule 16: Applicants should provide valid identity and address documents before final approval.

Rule 17: Income documents such as salary slips, bank statements, or business financial records may be required.

Rule 18: The requested loan amount should be consistent with the applicant's income and repayment capacity.

Rule 19: Applicants requesting an unusually high loan amount compared with their income should undergo additional risk assessment.

Rule 20: A loan application may be rejected when the applicant's verified income is insufficient to support the proposed EMI.

Rule 21: Existing credit card and loan obligations should be considered when calculating repayment capacity.

Rule 22: Applicants with multiple recently opened credit accounts may require additional credit assessment.

Rule 23: Consistent repayment history improves the applicant's overall credit profile.

Rule 24: Frequent missed or delayed payments may negatively affect loan eligibility.

Rule 25: Applications containing inconsistent income or employment information should be flagged for verification.

Rule 26: If submitted documents cannot be verified, the application should be placed on hold until verification is completed.

Rule 27: Applicants with stable income, strong credit history, and manageable existing debt may qualify for faster processing.

Rule 28: Loan approval should consider multiple factors rather than relying only on the credit score.

Rule 29: If the applicant does not meet the minimum eligibility criteria, the application should not proceed to final approval.

Rule 30: Final approval should be based on verified applicant information and the overall risk assessment.
"""




splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=30
)

documents = splitter.create_documents([loan_rules])

print("Loan rules loaded successfully!")
print("Number of document chunks:", len(documents))




def ask(question, keywords):

    print("=" * 70)
    print("QUESTION:", question)
    print("=" * 70)

    found = False

    for doc in documents:

        # Check whether any keyword exists in the rule
        if any(word.lower() in doc.page_content.lower()
               for word in keywords):

            print("ANSWER:")
            print(doc.page_content)
            print()

            found = True

    if not found:
        print("ANSWER: No matching loan approval rule found.")

    print()




ask(
    "What is the minimum age required for a personal loan?",
    ["21", "age"]
)

ask(
    "What credit score is considered low risk?",
    ["750", "credit score"]
)

ask(
    "Can an applicant with a credit score of 680 get a loan?",
    ["650", "749"]
)

ask(
    "How much existing debt can an applicant have?",
    ["40%", "debt", "income"]
)

ask(
    "What happens if the applicant has a credit score below 650?",
    ["below 650", "enhanced review"]
)

ask(
    "What documents are needed for loan approval?",
    ["identity", "address", "documents"]
)

ask(
    "What happens if an applicant has missed loan payments?",
    ["missed", "delayed", "payments"]
)

ask(
    "Can someone who has worked for only 3 months apply for a loan?",
    ["6 months", "employment"]
)

ask(
    "What happens if the income information is inconsistent?",
    ["inconsistent", "income", "verification"]
)

ask(
    "What factors are considered before approving a loan?",
    ["multiple factors", "risk assessment"]
)




print("\n")
print("=" * 70)
print("        🏦 LOAN APPROVAL RULE CHATBOT 🏦")
print("=" * 70)
print("Ask questions about loan approval rules.")
print("Type 'exit' to stop.")
print("=" * 70)


# Keyword dictionary
keyword_map = {

    "age": [
        "age",
        "old",
        "young"
    ],

    "income": [
        "income",
        "salary",
        "salary slip",
        "earn"
    ],

    "credit": [
        "credit",
        "credit score",
        "cibil",
        "score"
    ],

    "debt": [
        "debt",
        "loan obligation",
        "existing loan",
        "existing debt"
    ],

    "employment": [
        "employment",
        "job",
        "employed",
        "work",
        "experience"
    ],

    "documents": [
        "document",
        "identity",
        "address",
        "proof",
        "bank statement"
    ],

    "default": [
        "default",
        "missed payment",
        "late payment",
        "delayed payment"
    ],

    "loan amount": [
        "loan amount",
        "amount",
        "borrow"
    ],

    "emi": [
        "emi",
        "repayment",
        "repay"
    ],

    "approval": [
        "approval",
        "approve",
        "eligible",
        "eligibility",
        "qualify"
    ]
}


# Interactive loop

while True:

    question = input("\nEnter your loan question: ")

    if question.lower() == "exit":
        print("\nThank you for using the Loan Approval RAG System!")
        break

    # Find relevant keywords
    matched_keywords = []

    for category, words in keyword_map.items():

        for word in words:

            if word.lower() in question.lower():
                matched_keywords.append(word)

    # Remove duplicates
    matched_keywords = list(set(matched_keywords))

    if len(matched_keywords) == 0:

        print("\nNo relevant loan category detected.")
        print("Try asking about:")
        print("- Credit score")
        print("- Income")
        print("- Age")
        print("- Employment")
        print("- Debt")
        print("- Documents")
        print("- EMI")
        print("- Loan approval")

    else:

        print("\nSearching loan approval rules...\n")

        found = False

        for doc in documents:

            if any(
                keyword.lower() in doc.page_content.lower()
                for keyword in matched_keywords
            ):

                print("📌 Retrieved Rule:")
                print(doc.page_content)
                print()

                found = True

        if not found:

            print("No matching rule was found in the knowledge base.")
```
