```python
#1. FOOD DELIVERY EXPERIENCE SENTIMENT ANALYZER
#Problem Statement: A food delivery platform wants to analyze customer comments about delivery speed, food temperature, packaging, and rider behavior.

from transformers import pipeline

sentiment = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

reviews = [
    "My food arrived hot and much earlier than expected.",
    "The delivery person was rude and unhelpful.",
    "The packaging was excellent and nothing spilled.",
    "My order was delayed and the food was cold.",
    "The rider was polite and delivered everything safely."
]

results = sentiment(reviews)

for review, result in zip(reviews, results):
    print("Review:", review)
    print("Sentiment:", result["label"])
    print("Confidence:", round(result["score"], 3))
    print("-" * 50)



#2. ONLINE EDUCATION COURSE FEEDBACK ANALYZER
#Problem Statement: An online learning platform wants to understand student opinions about recorded lectures, assignments, and teaching quality.

import pandas as pd
from transformers import pipeline

sentiment = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

feedback = [
    "The instructor explains difficult topics very clearly.",
    "The recorded lectures are confusing and poorly organized.",
    "The assignments helped me understand the subject.",
    "The course contains too many errors.",
    "I enjoyed the interactive learning activities.",
    "The lessons are boring and difficult to follow."
]

results = sentiment(feedback)

df = pd.DataFrame({
    "Student_Feedback": feedback,
    "Sentiment": [r["label"] for r in results],
    "Confidence": [round(r["score"], 3) for r in results]
})

total = len(df)
positive = (df["Sentiment"] == "POSITIVE").sum()

print(df.to_string(index=False))
print("\\nTotal feedback:", total)
print("Positive feedback:", positive)
print("Negative feedback:", total - positive)
print("Positive percentage:", round(positive / total * 100, 2), "%")



#3. HOSPITAL PATIENT EXPERIENCE MONITOR
#Problem Statement: A hospital administration team wants to analyze patient comments about appointment scheduling, waiting time, staff behavior, and cleanliness.

from transformers import pipeline

sentiment = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

patient_feedback = [
    "The nurses were caring and treated me with respect.",
    "I waited for hours before seeing the doctor.",
    "The hospital rooms were clean and comfortable.",
    "The receptionist was unhelpful and impatient.",
    "The appointment process was smooth and convenient.",
    "The waiting area was dirty and uncomfortable."
]

results = sentiment(patient_feedback)

print("PATIENT EXPERIENCE MONITOR")
print("=" * 65)

for feedback, result in zip(patient_feedback, results):
    if result["label"] == "NEGATIVE" and result["score"] >= 0.80:
        status = "REVIEW REQUIRED"
    elif result["label"] == "NEGATIVE":
        status = "CHECK FEEDBACK"
    else:
        status = "NO NEGATIVE FLAG"

    print("Feedback:", feedback)
    print("Sentiment:", result["label"])
    print("Confidence:", round(result["score"], 3))
    print("Status:", status)
    print("-" * 50)


#4. TRAVEL AND HOTEL EXPERIENCE ANALYZER
#Problem Statement: A travel agency wants to evaluate tourist feedback about hotel accommodation, sightseeing, transport, and travel arrangements.

import pandas as pd
from transformers import pipeline

sentiment = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

data = {
    "Category": [
        "Hotel", "Transport", "Sightseeing",
        "Hotel", "Transport", "Sightseeing"
    ],
    "Feedback": [
        "The hotel room was spacious and beautiful.",
        "The bus was late and extremely uncomfortable.",
        "The mountain views were absolutely wonderful.",
        "The hotel staff were unfriendly and careless.",
        "The driver was professional and helpful.",
        "The tour was disappointing and poorly planned."
    ]
}

df = pd.DataFrame(data)

results = sentiment(df["Feedback"].tolist())

df["Sentiment"] = [r["label"] for r in results]
df["Confidence"] = [round(r["score"], 3) for r in results]

print("TRAVEL EXPERIENCE ANALYSIS")
print(df.to_string(index=False))

print("\\nNEGATIVE FEEDBACK BY CATEGORY")
negative = df[df["Sentiment"] == "NEGATIVE"]

if not negative.empty:
    print(
        negative.groupby("Category").size().sort_values(
            ascending=False
        )
    )
else:
    print("No negative feedback detected.")



#5. PUBLIC TRANSPORT PASSENGER SATISFACTION MONITOR
#Problem Statement: A public transport authority wants to understand passenger opinions about bus punctuality, seating, cleanliness, and staff conduct.

import pandas as pd
from transformers import pipeline

sentiment = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

passenger_feedback = [
    "The bus arrived on time and the seats were comfortable.",
    "The bus was overcrowded and very dirty.",
    "The conductor was helpful and polite.",
    "The service is unreliable and buses are always late.",
    "The new bus route makes my journey much easier.",
    "The seats were broken and uncomfortable."
]

results = sentiment(passenger_feedback)

df = pd.DataFrame({
    "Passenger_Feedback": passenger_feedback,
    "Sentiment": [r["label"] for r in results],
    "Confidence": [round(r["score"], 3) for r in results]
})

print("PUBLIC TRANSPORT FEEDBACK REPORT")
print(df.to_string(index=False))

print("\\nSENTIMENT SUMMARY")
print(df["Sentiment"].value_counts())

negative_feedback = df[df["Sentiment"] == "NEGATIVE"]

print("\\nFEEDBACK REQUIRING ATTENTION")
for feedback in negative_feedback["Passenger_Feedback"]:
    print("-", feedback)



#6. MOBILE NETWORK SERVICE QUALITY ANALYZER
#Problem Statement: A telecommunications company wants to monitor customer opinions about internet speed, network coverage, call quality, and service interruptions.

import pandas as pd
from transformers import pipeline

sentiment = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

network_feedback = [
    "The internet speed is excellent even during peak hours.",
    "Calls keep dropping in my area.",
    "The network coverage has improved significantly.",
    "Mobile data is extremely slow and frustrating.",
    "Customer service resolved my network issue quickly.",
    "I cannot get a stable signal inside my home.",
    "Video streaming works smoothly on this network."
]

results = sentiment(network_feedback)

df = pd.DataFrame({
    "Network_Feedback": network_feedback,
    "Sentiment": [r["label"] for r in results],
    "Confidence": [round(r["score"], 3) for r in results]
})

total = len(df)
negative_count = (df["Sentiment"] == "NEGATIVE").sum()
positive_count = (df["Sentiment"] == "POSITIVE").sum()

negative_percentage = (
    negative_count / total * 100 if total > 0 else 0
)

print("MOBILE NETWORK QUALITY REPORT")
print("=" * 65)
print(df.to_string(index=False))

print("\\nFINAL SUMMARY")
print("Total comments:", total)
print("Positive comments:", positive_count)
print("Negative comments:", negative_count)
print(
    "Negative feedback percentage:",
    round(negative_percentage, 2),
    "%"
)

print("\\nNETWORK ISSUES TO REVIEW")
for _, row in df[df["Sentiment"] == "NEGATIVE"].iterrows():
    print("-", row["Network_Feedback"])
```
