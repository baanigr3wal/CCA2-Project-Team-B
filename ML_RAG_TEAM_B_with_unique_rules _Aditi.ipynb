```python
college_rules = """
Students must maintain at least 75% attendance.

There are two internal tests during the semester.

Students can borrow up to three books from the library.

Students must carry their valid examination ID card.

Assignments must be submitted before the announced deadline.

Students need atleast 15 days preparation leave before final exams.

Late fee of Rs 50 per day applies after the fee due date.

Students with medical leave must submit a doctor certificate within 3 days.

The library remains open from 9 AM to 6 PM on weekdays.

Mobile phones are not allowed inside the examination hall.
"""

from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=20)
documents = splitter.create_documents([college_rules])

def ask(question, keywords):
    print("Q:", question)
    found = False
    for doc in documents:
        if any(w.lower() in doc.page_content.lower() for w in keywords):
            print("A:", doc.page_content)
            found = True
    if not found:
        print("A: No matching rule found.")
    print()

ask("How many days is preparation leave?", ["preparation"])
ask("What is the late fee?", ["late fee", "fee"])
ask("When is the library open?", ["library"])
ask("Can I use my phone in exams?", ["mobile", "phone"])
```

```python
# Unique Rule Set: College Fest and Student Club Guidelines
# This section uses a separate rule collection so it does not mix with the original college rules.

club_event_rules = """
Event proposals must be submitted to the Student Council at least 7 days before the planned event.
Promotional posters must be approved by the faculty coordinator before they are published or displayed.
Venue bookings must be requested through the campus event portal and are confirmed only after approval.
Sponsorship offers must be reviewed by the club's faculty advisor before any agreement is made.
Every event budget must include item-wise estimates and be approved before purchases are made.
Purchases above Rs 2000 require at least two vendor quotations for comparison.
Original bills and expense receipts must be submitted to the club treasurer within 5 working days after the event.
Borrowed sound, lighting, and presentation equipment must be returned to the media store within 24 hours after use.
Names of outside guests must be shared with campus security at least 48 hours before the event.
Food stalls at student events require written permission from the campus canteen office.
Each event must assign one volunteer to coordinate first aid and one volunteer to manage crowd movement.
Event music and amplified sound must stop by 7 PM unless special written permission has been granted.
"""

from langchain_text_splitters import RecursiveCharacterTextSplitter

club_splitter = RecursiveCharacterTextSplitter(chunk_size=180, chunk_overlap=30)
club_documents = club_splitter.create_documents([club_event_rules])

def ask_club(question, keywords):
    print("Q:", question)
    matches = []
    for doc in club_documents:
        if any(word.lower() in doc.page_content.lower() for word in keywords):
            if doc.page_content not in matches:
                matches.append(doc.page_content)
    if matches:
        for match in matches:
            print("A:", match)
    else:
        print("A: No matching rule found.")
    print()

# Sample questions
ask_club("How early should an event proposal be submitted?", ["Event proposals"])
ask_club("Who approves promotional posters?", ["Promotional posters"])
ask_club("How do we book a venue?", ["Venue bookings"])
ask_club("What is needed for purchases above Rs 2000?", ["Purchases above Rs 2000"])
ask_club("When should receipts be submitted?", ["expense receipts"])
ask_club("How soon must borrowed equipment be returned?", ["Borrowed sound"])
ask_club("When should guest names be shared with security?", ["outside guests"])
ask_club("Who gives permission for food stalls?", ["Food stalls"])
ask_club("Who coordinates first aid and crowd movement?", ["first aid"])
ask_club("When must event music stop?", ["Event music"])
```
